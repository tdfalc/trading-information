```python
import numpy as np

def riemann_sum_midpoint(f, a, b, n):
    """
    Compute the Riemann sum using the midpoint rule.
    
    Parameters:
    f : function
        The function to integrate
    a : float
        Lower bound of integration
    b : float
        Upper bound of integration
    n : int
        Number of subintervals
    
    Returns:
    float
        Approximation of the integral
    """
    x = np.linspace(a, b, n+1)
    dx = (b - a) / n
    x_mid = (x[:-1] + x[1:]) / 2
    return np.sum(f(x_mid) * dx)

# Example usage:
def f(x):
    return x**2

result = riemann_sum_midpoint(f, 0, 1, 1000)
print(f"Approximate integral: {result}")
```
