```python
import numpy as np
from scipy.optimize import minimize
```

```python
x0 = np.random.uniform(size=4)
v = 0.4
# def objective(x):
#     pgG, pgN, pnG, pnN = x
#     return -(v * pgG + (1 - v) * pgN ) 

def objective(x):
    pgG, pgN, pnG, pnN = x
    return -(v * (pgG + 0) + (1-v) * (pgN + 0))

def objective(x):
    pgG, pgN, pnG, pnN = x
    return -(v * (pgG * 1 + pnG * 0) + (1-v) * (pgN * 1 + pnN * 0))

def obed1(x):
    pgG, pgN, pnG, pnN = x
    return v * pgG - (1 - v)  *pgN

def obed2(x):
    pgG, pgN, pnG, pnN = x
    return (1-v) * pnN - v * pnG

def prop1(x):
    pgG, pgN, pnG, pnN = x
    return pgG + pnG - 1

def prop2(x):
    pgG, pgN, pnG, pnN = x
    return pgN + pnN - 1


# Define the constraints
constraints = [
    {'type': 'ineq', 'fun': obed1},
    {'type': 'ineq', 'fun': obed2},
    {'type': 'eq', 'fun': prop1},
    {'type': 'eq', 'fun': prop2}
]

# Solve the optimization problem
result = minimize(objective, x0, constraints=constraints, bounds=[(0, 1), (0, 1), (0, 1), (0, 1)])
pgG, pgN, pnG, pnN = result.x
print(pgG, pnN)
```
