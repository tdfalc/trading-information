```python
import numpy as np

def find_nash_equilibria(utility_player1, utility_player2):
    nash_equilibria = []

    # Iterate over all possible strategy profiles
    for i in range(2):  # Player 1's strategies (row)
        for j in range(2):  # Player 2's strategies (column)
            # Check if player 1's strategy i is a best response to player 2's strategy j
            best_response_player1 = (utility_player1[i, j] >= utility_player1[1 - i, j])
            
            # Check if player 2's strategy j is a best response to player 1's strategy i
            best_response_player2 = (utility_player2[i, j] >= utility_player2[i, 1 - j])
            
            # If both players are playing a best response, it's a Nash equilibrium
            if best_response_player1 and best_response_player2:
                nash_equilibria.append((i, j))
    
    return nash_equilibria

def find_nash_equilibria_with_dominant_strategies(utility_player1, utility_player2):
    nash_equilibria = []

    # Iterate over all possible strategy profiles
    for i in range(2):  # Player 1's strategies (row)
        for j in range(2):  # Player 2's strategies (column)
            # Check if player 1's strategy i is a best response to player 2's strategy j
            best_response_player1 = (utility_player1[i, j] >= utility_player1[1 - i, j])
            
            # Check if player 2's strategy j is a best response to player 1's strategy i
            best_response_player2 = (utility_player2[i, j] >= utility_player2[i, 1 - j])
            
            # If both players are playing a best response, it's a Nash equilibrium
            if best_response_player1 and best_response_player2:
                # Check if Player 1's strategy i is a dominant strategy
                dominant_player1 = all(utility_player1[i, k] >= utility_player1[1 - i, k] for k in range(2))
                
                # Check if Player 2's strategy j is a dominant strategy
                dominant_player2 = all(utility_player2[k, j] >= utility_player2[k, 1 - j] for k in range(2))
                
                # Append the equilibrium with dominance information
                nash_equilibria.append((i, j, dominant_player1, dominant_player2))
    
    return nash_equilibria


# Example usage:



utility_player1 = lambda tau: np.array([
    [1-tau, 1], 
    [-tau, 0]
    ])

utility_player2 = lambda tau: np.array([
        [1-tau, -tau], 
    [1, 0]
    ])


tau = 32

        
ne = find_nash_equilibria(utility_player1(tau), utility_player2(tau))
ds = find_nash_equilibria_with_dominant_strategies(utility_player1(tau), utility_player2(tau))

print("--")
print(np.round(tau, 3), "Nash Equilibria:", ne)
print(np.round(tau, 3), "Nash Equilibria:", ds)
```
